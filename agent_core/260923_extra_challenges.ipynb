```python
#@title 정답 챌린지 5 { display-mode: "form" }
from collections import Counter
ip_counts = Counter(["10.0.3.21", "10.0.3.21", "10.0.3.21", "10.0.9.8", "10.0.8.15", "10.0.9.8", "10.0.3.21", "10.0.7.7"])
top_ip = ""
top_count = 0
for ip, count in ip_counts.items():
    if count > top_count:
        top_ip = ip
        top_count = count
print(f"가장 많은 실패: {top_ip} · {top_count}회")
```
